# Esercitazione sulla Rappresentazione Numerica

## Esercizio 1

Creare uno script python che:
1. chieda in input 2 valori interi;
2. stampi i valori in rappresentazione binaria ed esadecimale;
3. satmpi il risultato delle operazioni AND e OR bitwise sia in rappresentazione binaria che esadecimale.

## Esercizio 2

Creare uno script python che:
1. chieda in input un valore intero e il numero di bit cui applicare delle operazioni di bit shift;
2. stampi il valore intero in rappresentazione binaria ed esadecimale;
3. satmpi il risultato delle operazioni di bit shift (`<<` e`>>`) sia in rappresentazione binaria che esadecimale.
